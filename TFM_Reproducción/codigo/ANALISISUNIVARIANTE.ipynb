```python
def obtener_prevalencia(dataframe):
    prevalencia = dataframe["specific_treatment_type"].value_counts(normalize=True)*100
    return prevalencia
```

```python
def contar_casos(dataframe):
   casos = dataframe["specific_treatment_type"].value_counts()
   return casos 
```

```python
def obtener_tasanacido(dataframe):
    tasa_nacido = dataframe.groupby("specific_treatment_type")["live_birth_occurrence"].mean()*100
    return tasa_nacido
```

```python
def contar_nacidos(dataframe):
    nacidos = dataframe.groupby("specific_treatment_type")["live_birth_occurrence"].sum()
    return nacidos
```

```python
def chi_cuadrado(dataframe):
    from scipy.stats import chi2_contingency
    tabla_exito = pd.crosstab(dataframe["specific_treatment_type"],dataframe["live_birth_occurrence"])
    chi2, p_valor, dof, expected = chi2_contingency(tabla_exito)
    if p_valor < 0.05 :
         print(f"RESULTADO SIGNIFICATIVO, p: {p_valor:.4f}")
    else:
         print(f"RESULTADO NO SIGNIFICATIVO, p: {p_valor:.4f}")
```

```python
def odds_ratio(dataframe):
    import statsmodels.api as sm
    tabla_exito = pd.crosstab(dataframe["specific_treatment_type"],dataframe["live_birth_occurrence"])
    tabla_reordenada = tabla_exito.loc[["IVF", "ICSI"],[1, 0]]
    table = sm.stats.Table2x2(tabla_reordenada.values)
    or_valor = table.oddsratio
    ic_bajo, ic_alto =table.oddsratio_confint()
    return print(F"ODDS RATIO: {or_valor:.4f}\nINTERVALO CONFIANZA 95% : [{ic_bajo:.4f} - {ic_alto:.4f}]")
```
