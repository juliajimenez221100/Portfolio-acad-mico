```python
import pandas as pd
```

```python
df_completo = pd.read_csv(r"C:\Users\Usuario\Downloads\Datos_TFM_NUEVOS2304.csv")
```

```python
df_nofactormasc = df_completo[df_completo["causes_of_infertility_-_male_factor"]==0]
```

```python
df_nodonantes = df_nofactormasc[df_nofactormasc["sperm_source"]== "Partner"]
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
formula = """live_birth_occurrence ~ 
             C(specific_treatment_type, Treatment(reference='ICSI')) + 
             C(patient_age_grouped, Treatment(reference='<35')) + 
             C(eggs_collected_grouped, Treatment(reference='respuesta_optima')) + 
             C(Previous_ivf_cycles_grouped, Treatment(reference='primer_intento'))"""


modelo = smf.logit(formula, data=df_nodonantes).fit()
print(modelo.summary())


resultados_or = np.exp(modelo.params)
intervalos_conf = np.exp(modelo.conf_int())


tabla_final = pd.concat([resultados_or, intervalos_conf], axis=1)
tabla_final.columns = ['OR', 'IC 95% Bajo', 'IC 95% Alto']
print(tabla_final)
```
